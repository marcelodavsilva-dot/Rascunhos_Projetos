```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1.Carregando o conjunto de dados "artificial1d.csv".
df = np.loadtxt("artificial1d.csv", delimiter= ",", skiprows= 1)
```

```python
# 2.Embaralhando o conjunto de dados (Evitas vieses).
np.random.seed(10)
np.random.shuffle(df)
```

```python
# 3. Criação do vetor X (com a variável de entrada) e do vetor Y (com a variável de saída) | arrays 2D.
X = df[:,[0]]
y = df[:,[1]]
```

```python
# 4. Criação das variáveis polinomiais X², X³ e X^4 e adição à matriz X
var_pol = np.c_[X, X**2, X**3, X**4]
```

```python
# 5. Separação do conjunto de treino (Aprox.70% do conj. dados) e do conjunto de teste (Aprox.30% do conj. dados).
X_treino = var_pol[:20]
X_teste = var_pol[20:]
y_treino = y[:20]
y_teste = y[20:]
```

```python
# 6. Treinamento do modelo de regressão linear regularizado.
def regressao (var_pol, y, L):

    X1 = np.c_[np.ones(var_pol.shape[0]), var_pol]
    l = np.eye(X1.shape[1])
    l[0,0] = 0
    L_ = l * L
    w = (np.linalg.inv( (X1.T) @ X1+L_)) @ ((X1.T) @ y)

    return (w)
```

```python
def predict_multivariate(X,w):

    c1 = np.c_[np.ones(X.shape[0]), X]

    y_predict = c1 @ w

    return y_predict
```

```python
# 7. Reporte do MSE, MAPE e R² para o conjunto de teste.
def MSE_teste (y_teste, y_predict):
    mse_teste = np.mean ((y_teste - y_predict)**2)

    return mse_teste

def MAPE (y_teste, y_predict):
    mape = 100*(np.mean(np.abs((y_teste - y_predict)/y_teste)))

    return mape

def R2 (y_teste, y_predict):
    r2 = 1 - np.sum((y_teste - y_predict)**2) / np.sum((y_teste-(np.mean(y_teste)))**2)

    return r2
```

```python
# 8. Repetição (loop) dos passos 6 e 7 utilizando os seguintes valores para Lambda:
lambdas = (0, 0.001, 0.01, 0.1, 1, 10, 100)

for L in lambdas:
  print (f"\n", "Lambda:", L, f"\n")

  w = regressao(X_treino, y_treino, L)
  print ("W:", w, f"\n")

  y_predict = predict_multivariate(X_teste, w)
  print ("Valores predição (TESTE) com Lambda:", L, y_predict, f"\n")

  mse_teste = MSE_teste (y_teste, y_predict)
  print ("MSE:", mse_teste, f"\n")

  mape = MAPE(y_teste, y_predict)
  print ("MAPE:", mape, f"\n")

  r2 = R2 (y_teste, y_predict)
  print ("R2:", r2, f"\n")

  plt.scatter(X_teste[:, 0], y_teste, color='blue', label='Dados reais (TESTE)')

  reord_indices = np.argsort(X_teste[:, 0])
  X_reord = X_teste[reord_indices, 0]
  y_reord = y_predict[reord_indices]
  plt.plot(X_reord, y_reord, color = "red", label=f'Previsão Lambda={L}')

  plt.xlabel('X')
  plt.ylabel('y')
  plt.title('Previsão')
  plt.legend()
  plt.grid(True)
  plt.show()
```
