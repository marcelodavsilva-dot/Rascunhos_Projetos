```python
#4
var_poli = np.c_[X, X**2, X**3, X**4]
print (var_poli.shape)
```

```python
X_treino = var_poli[:20]
X_teste = var_poli[20:]
y_treino = y[:20]
y_teste = y[20:]
```

```python
def regressao_mult_reg (var_poli, y, L):

    X1 = np.c_[np.ones(var_poli.shape[0]), var_poli]
    l = np.eye(X1.shape[1])
    l[0,0] = 0
    L_ = l * L
    w = (np.linalg.inv( (X1.T) @ X1+L_)) @ ((X1.T) @ y)

    return (w)
```

```python
w = regressao_mult_reg (X_treino, y_treino, 1)
print (w)
```
