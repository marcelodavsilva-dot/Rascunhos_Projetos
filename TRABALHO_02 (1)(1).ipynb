```python
import numpy as np
```

```python
#1
df = np.loadtxt("artificial1d.csv", delimiter= ",", skiprows= 1)
```

```python
#2
np.random.seed(3)
np.random.shuffle(df)
```

```python
#3
X = df[:,[0]]
y = df[:,[1]]
```

```python
#4
var_pol = np.c_[X, X**2, X**3, X**4]
print (var_pol.shape)
```

```python
#5
X_treino = var_pol[:20]
X_teste = var_pol[20:]
y_treino = y[:20]
y_teste = y[20:]
```

```python
#6
def regressao (var_pol, y, L):

    X1 = np.c_[np.ones(var_pol.shape[0]), var_pol]
    l = np.eye(X1.shape[1])
    l[0,0] = 0
    L_ = l * L
    w = (np.linalg.inv( (X1.T) @ X1+L_)) @ ((X1.T) @ y)

    return (w)
```

```python
w = regressao(X_treino, y_treino, 0)
print (w)
```

```python
w = regressao(X_teste, y_teste, 0)
print (w)
```

```python
def predict_multivariate(X,w):

    c1 = np.c_[np.ones(X.shape[0]), X]

    y_predict = c1 @ w

    return y_predict
```

```python
y_predict = predict_multivariate(X_treino, w)
print (y_predict)
```

```python
y_predict = predict_multivariate(X_teste, w)
print (y_predict)
```

```python
#7
def MSE_teste (y_teste, y_predict):
    mse_teste = np.mean ((y_teste - y_predict)**2)

    return mse_teste

def MAPE (y_teste, y_predict):
    mape = 100*(np.mean((y_teste - y_predict)/y_teste))

    return mape

def R2 (y_teste, y_predict):
    r2 = 1 - np.sum((y_teste - y_predict)**2) / np.sum((y_teste-(np.mean(y_teste)))**2)

    return r2
```

```python
mse_teste = MSE_teste (y_teste, y_predict)
print ("MSE:", mse_teste)
```

```python
mape = MAPE(y_teste, y_predict)
print ("MAPE:", mape)
```

```python
r2 = R2 (y_teste, y_predict)
print ("R2:", r2)
```
