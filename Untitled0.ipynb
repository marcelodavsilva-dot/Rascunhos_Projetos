```python
import numpy as np
```

```python
data = np.loadtxt("boston_housing.csv", delimiter= ",", skiprows = 1)
```

```python
X = data[:,1]
y = data[:,3]

def standardize(v):
  y = (v - np.mean(v))/np.std(v)

  return y
```

```python
def regressao_linear (X, y):

  x_ = np.mean(X)
  y_ = np.mean(y)

  w1 = (np.sum((X-x_)*(y-y_)))/np.sum((X-x_)**2)
  w0 = y_ - (w1*x_)

  return w0, w1
```

```python
w0, w1 = regressao_linear(X, y)
print (w0, w1)
```
