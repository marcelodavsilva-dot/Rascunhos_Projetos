```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Synthetic_Financial.csv")
```

```python
df.describe()
```

```python
print ("Contagem de 'isFraud':")
print (df['isFraud'].value_counts())
```

```python
print ("Proporção de 'isFraud':")
print (df['isFraud'].value_counts(normalize=True))
```

```python
plt.figure(figsize=(6, 4))
sns.set_style('whitegrid')
sns.countplot(data = df, x = 'isFraud', hue = 'isFraud')
plt.yscale('log')
plt.title("Distribuição entre transações Fraudulentas x Normais", fontsize = 15, y = 1.03)
plt.xlabel("Identificação")
plt.ylabel('Contagem')
plt.legend(labels = ["0 - NORMAL", "1 - FRAUDE"])
plt.show()
```

```python
# Em quais tipos de transação a fraude ocorre?

plt.figure(figsize=(6, 4))
sns.set_style('whitegrid')
sns.countplot(data = df, x = 'type', hue = 'isFraud')
plt.yscale('log')
plt.title('Contagem de Transações por Tipo e Classe (Fraude/Normal)', fontsize = 15, y = 1.05)
plt.xlabel('Tipo de Transação')
plt.ylabel('Contagem')
plt.legend(title='É Fraude?', labels=['0 - Normal', '1 - Fraude'])
plt.show()
```

```python
# Verificado que as fraudes só ocorrem nos tipos "TRANSFER" e "CASH_OUT".

# Visualização da distribuição de valores.

plt.figure(figsize=(10, 6))
sns.set_style('whitegrid')
sns.histplot(data=df, x="amount", bins=200, log_scale = True) # Aplicação do "log_scale" para uma melhor visualização da distribuição.
plt.title("Distribuição dos Valores das Transações - (escala logarítmica)", fontsize = 18, y = 1.03)
plt.xlabel("Valor (Amount) - Em log")
plt.ylabel("Frequência")
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.set_style('whitegrid')
sns.boxplot(data=df, x='isFraud', y='amount')
plt.title('Comparação de Valores (Amount) entre as Classes')
plt.yscale('log')
plt.xlabel('Classes: (0 = Normal, 1 = Fraude)')
plt.ylabel('Valor (Amount)')
plt.show()
```
