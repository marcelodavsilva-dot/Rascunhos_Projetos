```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(2)
n = 200

interv_idade = np.random.uniform(0, 100, n).astype(int)
df = pd.DataFrame ({'idade': interv_idade})

criança = df["idade"] < 18
adulto = (df["idade"] >= 18) & (df["idade"] < 60)
idoso = df["idade"] >= 60

Criança = np.sum(criança)
print(Criança)

Adulto = np.sum(adulto)
print (Adulto)

Idoso = np.sum(idoso)
print (Idoso)

labels = ["0 > 18", "18 > 59", "60+"]
values = [Criança, Adulto, Idoso]

plt.bar(labels, values, color = "green", edgecolor = "black")
plt.xticks (labels)
plt.ylabel ("Quantidade")
plt.title ("Distribuição de Idade")
plt.show()
```
