```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
x = np.array([1.33,2.57,3.99,4.43,7.26,8.45]).reshape(-1,1)
y = np.array([5,6,7,8,14,300])

modelo = LinearRegression()
modelo.fit(x, y)

print ("Coeficiente (a):", modelo.coef_[0])
print ("Intercepto(b):", modelo.intercept_)

y_pred = modelo.predict(x)

plt.scatter(x, y, color = "blue", label = "Dados reais")
plt.plot(x, y_pred, color = "red", label = "Regressão linear")
plt.xlabel("Variável independente (x)")
plt.ylabel("Variável dependente (y)")
plt.title("Regressão linear")
plt.legend()
plt.show()

```

```python
data = np.loadtxt ("boston_housing.csv", delimiter = ",", skiprows = 1)

X = data[:,1]
y = data[:,3]
```

```python
data
```

```python
def regressao_linear (X, y, alpha, n_epocas):

  b0 = 0
  b1 = 0

  for i in range (n_epocas):
    y_ = b1*X + b0
    e = y - y_
    soma_ex = np.sum(e*X)
    soma_e = np.sum(e)

    b0 = b0 + (alpha*(1/n_epocas))*soma_e
    b1 = b1 + (alpha*(1/n_epocas))*soma_ex

  return b0, b1


```

```python
b0, b1 = regressao_linear (X, y, 0.01, 1000)
print (b0, b1)
```

```python
y_ = b1*X + b0

plt.scatter(X, y, color = "pink", label = "Dados reais")
plt.plot(X, y_, color = "red", label = "Regressão linear")
plt.xlabel("Variável independente (x)")
plt.ylabel("Variável dependente (y)")
plt.title("Regressão linear")
plt.legend()
plt.show()
```

```python
data = np.loadtxt("TESTE - REGRESSAO_INFLACAO (4).csv", delimiter = ",", skiprows = 1)
```

```python
X = data[:,1]
y = data[:,0]
```

```python
def regressao_linear_inflacao (X, y, alfa, n):
  b0 = 0
  b1 = 1

  for i in range (n):
    y_ = b1*X+b0
    e = y - y_
    soma_ex = np.sum(e*X)
    soma_e = np.sum(e)

    b0 = b0 + (alfa*(1/n))*soma_e
    b1 = b1 + (alfa*(1/n))*soma_ex

    return b0, b1
```

```python
b0, b1 = regressao_linear_inflacao(X, y, 0.00001, 10000000)
print (b0, b1)
```

```python
y_ = b1*X + b0

plt.scatter(X, y, color = "pink", label = "Dados reais")
plt.plot(X, y_, color = "red", label = "Regressão linear")
plt.xlabel("Variável independente (x)")
plt.ylabel("Variável dependente (y)")
plt.title("Regressão linear")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Gerando dados simulados
np.random.seed(42)
meses = pd.date_range(start='2015-01-01', periods=100, freq='M')
pib_mensal = np.random.normal(loc=0.3, scale=0.1, size=100)  # PIB mensal (variação em %)
ipca_mensal = 0.5 * pib_mensal + np.random.normal(loc=0, scale=0.1, size=100)  # IPCA influenciado pelo PIB

# Criando DataFrame
df = pd.DataFrame({
    'Data': meses,
    'PIB_Mensal': pib_mensal,
    'IPCA_Mensal': ipca_mensal
})

# Regressão linear
X = df[['PIB_Mensal']]
y = df['IPCA_Mensal']
modelo = LinearRegression()
modelo.fit(X, y)
y_pred = modelo.predict(X)

# Plotando os dados e a regressão
plt.figure(figsize=(10, 6))
plt.scatter(X, y, label='Dados reais', alpha=0.7)
plt.plot(X, y_pred, color='red', label='Regressão Linear')
plt.xlabel('PIB Mensal (%)')
plt.ylabel('IPCA Mensal (%)')
plt.title('Regressão Linear: Previsão do IPCA com base no PIB Mensal')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
