```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
tv = np.array([10, 12, 9,7])
radio = np.array([8, 6, 9, 13])
jornal = np.array([12, 8, 5, 10])
vendas = np.array([12, 10, 11, 9])
```

```python
X = tv
y = vendas

def reg_lin_sim (X,y):

  x_ = np.mean(X)
  y_ = np.mean(y)

  w1 = (np.sum((X-x_)*(y-y_)))/np.sum((X-x_)**2)
  w0 = y_ - (w1*x_)

  return w0, w1
```

```python
w0, w1 = reg_lin_sim (X, y)
print (w0, w1)
```
