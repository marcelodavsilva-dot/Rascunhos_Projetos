```python
import numpy as np

data = np.loadtxt("boston_housing.csv", delimiter=",", skiprows=1)

v = data[:,[1,3]]
```

```python
def standardize (v):

    data_padr = np.zeros(v.shape)
    
    media0 = np.mean(v[:,0])
    desv0 = np.std(v[:,0])
    data_padr[:,0] = (v[:,0] - media0) / desv0

    media1 = np.mean(v[:,1])
    desv1 = np.std(v[:,1])
    data_padr[:,1] = (v[:,1] - media1) / desv1

    return (data_padr)
```

```python
data_padr = standardize(v)

X_true = data_padr[:,0]
y_true = data_padr[:,1]
```

```python
def regressao_linear(X_true, y_true):
    X_ = np.mean(X_true)
    y_ = np.mean(y_true)

    w1 = (np.sum((X_true-X_)*(y_true-y_)))/np.sum((X_true-X_)**2)
    w0 = y_- (w1*X_)

    return (w0, w1)
```

```python
w0, w1 = regressao_linear(X_true, y_true)
print(w0, w1)
```

```python
def predict_univariate(X, w0, w1):
    y_predict = w0 + (w1*X)

    return y_predict

def MSE (y, y_predict):
    mse = np.mean ((y - y_predict)**2)

    return mse

def R2 (y, y_predict):
    r2 = 1 - np.sum((y-y_predict)**2) / np.sum((y-(np.mean(y)))**2)

    return r2
```

```python
y_predict = predict_univariate(X_true, w0, w1)
print ("Predições:",y_predict)
```

```python
mse = MSE (y_true, y_predict)
print ("MSE:", mse)
```

```python
r2 = R2 (y_true, y_predict)
print ("R2:", r2)
```
