EXPERIMENTO 02 - Regressão Linear Univariada - Método "Gradiente descendente comum".

```python
# Carregamento do conjunto de dados "Boston House Price Dataset".

import numpy as np

data = np.loadtxt("boston_housing.csv", delimiter=",", skiprows=1)

v = data[:,[1,3]]
```

```python
# Implantação da função "standardize(v)" que recebe um vetor v e retorna o vetor v padronizado.

# Padronização da variável LSTAT como atributo preditor e da variável MEDV como atributo alvo. Obtendo o vetor as observações da coluna LSTAT e o vetor com as observações da coluna MEDV.

def standardize (v):

    data_padr = np.zeros(v.shape)

    media0 = np.mean(v[:,0])
    desv0 = np.std(v[:,0])
    data_padr[:,0] = (v[:,0] - media0) / desv0

    media1 = np.mean(v[:,1])
    desv1 = np.std(v[:,1])
    data_padr[:,1] = (v[:,1] - media1) / desv1

    return (data_padr)
```

```python
# Valores padronizados.

data_padr = standardize(v)

X_true = data_padr[:,0]
y_true = data_padr[:,1]
```

```python
# Treinamento do método de regressão linear univariada usando o método "Gradiente descendente comum".

def regressao_linear(X, y, alpha, n_epocas):

    w0 = 0
    w1 = 0

    # Utilizado p/ mostrar MSE e R2 com as predições já calculadas a cada rodada. 
    lista_mse = []
    lista_r2 = []

    for i in range (n_epocas):

        y_predict = w0 + w1*X
        e = y - y_predict
        w0 = w0 + alpha*np.mean(e)
        w1 = w1 + alpha*np.mean(e*X)

        # Utilizado p/ mostrar MSE e R2 com as predições já calculadas a cada rodada. 
        mse = np.mean ((y - y_predict)**2)
        r2 = 1 - np.sum((y - y_predict)**2) / np.sum((y-(np.mean(y)))**2)

        lista_mse.append(mse)
        lista_r2.append(r2)

    array_mse = np.array(lista_mse)
    array_r2 = np.array(lista_r2)

    return (w0, w1, array_mse, array_r2)
```

```python
# Coeficientes finais.

w0, w1, array_mse, array_r2 = regressao_linear(X_true, y_true, 1, 1000)
print (w0, w1)
```

```python
# Predições, Predições "despadronizadas", MSE e R2 obtidos com o modelo final.

def predict_univariate(X_true, w0, w1):
    y_predict = w0 + (w1*X_true)

    return y_predict

def predict_univariate_despadronizado(y_predict):

    y_predict_orig = y_predict * np.std(v[:,1]) + np.mean(v[:,1])

    return y_predict_orig

def MSE (y_true, y_predict):
    mse = np.mean ((y_true - y_predict)**2)

    return mse

def R2 (y_true, y_predict):
    r2 = 1 - np.sum((y_true-y_predict)**2) / np.sum((y_true-(np.mean(y_true)))**2)

    return r2
```

```python
# Predições, MSE e R² atualizados a cada rodada.

y_predict = predict_univariate(X_true, w0, w1)
print ("Predições:",y_predict)
print ("MSE atualizado:", array_mse)
print ("R2 atualizado:", array_r2)
```

```python
# Predição "despadronizada".

y_predict_orig = predict_univariate_despadronizado (y_predict)
print (y_predict_orig)
```

```python
# MSE

mse = MSE (y_true, y_predict)
print ("MSE:", mse)
```

```python
# R²

r2 = R2 (y_true, y_predict)
print ("R2:", r2)
```
