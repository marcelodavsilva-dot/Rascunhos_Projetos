# **Treinamento do modelo com o método "Regressão Logística".**

```python
# Importação de bibliotecas e APIs.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import accuracy_score
from sklearn.metrics import recall_score
from sklearn.metrics import precision_score
from sklearn.inspection import DecisionBoundaryDisplay
```

```python
# Importação do dataset "artificial2d.csv".

df = np.loadtxt("artificial2d.csv", delimiter=",", skiprows=1)
```

```python
print (df.shape)
```

```python
# Criação da matriz X com as variáveis de entrada e o vetor y com a variável de saída.

X = df[:,[0,1]]
y = df[:,2]
```

```python
# Separação do conjunto de treino e teste (70%/30%).

X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Padronização das variáveis.

padronizacao = StandardScaler().fit(X_treino)
X_treino_padronizado = padronizacao.transform(X_treino)
X_teste_padronizado = padronizacao.transform(X_teste)
```

```python
# Treinamento do modelo de regressão logística.

regressao_logistica = LogisticRegression(penalty = None)

regressao_logistica.fit(X_treino_padronizado, y_treino)
predicao = regressao_logistica.predict(X_teste_padronizado)
```

```python
# Reporte da acurácia.

accuracy_score(y_teste, predicao)
```

```python
# Reporte do Recall.

recall_score (y_teste, predicao)
```

```python
# Reporte da precision.

precision_score(y_teste, predicao)
```

```python
cm_regressao_logistica = confusion_matrix(y_teste, predicao)
```

```python
# Exibição da Matriz de Confusão.

disp_regressao_logistica = ConfusionMatrixDisplay(confusion_matrix=cm_regressao_logistica)
disp_regressao_logistica.plot(cmap='Blues')
plt.title("Matriz de confusão", fontsize=20, y=1.03)
plt.show()
```

```python
# Exibição do gráfico de Fronteira de Classificação.

DecisionBoundaryDisplay.from_estimator(regressao_logistica, X_teste_padronizado, response_method = "predict", cmap = plt.cm.coolwarm, alpha = 0.4, shading = 'auto')

plt.scatter(X_teste_padronizado[:, 0], X_teste_padronizado[:, 1], c = y_teste, cmap = plt.cm.coolwarm, edgecolor = 'k', s=70, linewidth = 1, label = "Observações")

plt.title("Regressão Logística - Fronteira de Classificação", fontsize=16, y=1.03)
plt.xlabel("Variável X₁", fontsize = 10)
plt.ylabel("Variável X₂", fontsize = 10)
plt.legend(fontsize=9)
plt.grid(True, linestyle = '--', alpha = 0.5)
plt.show()
```
