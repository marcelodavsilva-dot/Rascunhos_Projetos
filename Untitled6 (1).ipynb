```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
diab = datasets.load_diabetes(as_frame=True)
df = diab.frame.copy()
df.rename(columns={"target": "disease_progression"}, inplace=True)
```

```python
print(df[["bmi", "disease_progression"]].describe().round(3))
```

```python
sns.scatterplot(x="bmi", y="disease_progression", data=df)
plt.title("Dispersão: BMI vs Progressão da Doença")
plt.show()
```

```python
X = df["bmi"]
y = df["disease_progression"]
```

```python
def regressao_linear(X,y):
  y_ = np.mean(y)
  x_ = np.mean(X)
  b1 = np.sum((X-x_)*(y-y_))/np.sum((X-x_)**2)
  b0 = y_ - b1*x_
  return b0, b1
```

```python
b0, b1 = regressao_linear(X,y)
print(b0, b1)
```

```python
def predicao (X, b0, b1):
  y_pred = b0 + b1*X
  return y_pred
```

```python
def RSS (y, y_pred):
  rss = np.sum((y-y_pred)**2)
  return rss

def TSS (y):
  tss = np.sum((y-(np.mean(y)))**2)
  return tss
```

```python
y_pred = predicao(X, b0, b1)
print(y_pred)

rss = RSS(y, y_pred)
print("RSS:", rss)

tss = TSS(y)
print("TSS:", tss)

R2 = 1 - (rss/tss)
print("R2:", R2)
```

```python
sns.scatterplot(x="bmi", y="disease_progression", data=df)
plt.plot(X, y_pred, color="red")
plt.title("Regressão Linear: BMI / Progressão da Doença")
plt.show()
```
