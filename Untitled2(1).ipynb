```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
np.random.seed(2)
n = 100
```

```python
# Variáveis contínuas c/ outliers:

v_cont1 = np.random.normal(50, 10, n)
v_cont1[::10]*= 3
```

```python
v_cont2 = np.random.normal(30, 5, n)
v_cont2 [::15]*= 2
```

```python
# Criação de variáveis categóricas

cat = np.random.choice(['A', 'B', 'C'], n)
print (cat)
```

```python
# Criação de variáveis discretas

disc = np.random.poisson(3, n).astype(float)
print (disc)
```

```python
# Criação de variáveis binárias
bin = np.random.choice([0, 1], n)
print (bin)
```

```python
# Criação da variável texto
txt = np.random.choice (["Marcelo", "Laura", "Ricardo"], n)
print (txt)
```

```python
# Add valores null

v_cont1[15] = np.nan
v_cont2[10] = np.nan
cat[5] = np.nan
disc[10] = np.nan
```

```python
# Criando o dataframe

df = pd.DataFrame ({'v_cont1': v_cont1, 'v_cont2': v_cont2, 'cat': cat, 'disc': disc, 'bin': bin, 'txt': txt})
print ("shape:", df.shape)
print (df.head())
```

```python
# Medidas de tendência central

for i in df.columns:
  print(f"\n Variável: {i}")
  if pd.api.types.is_numeric_dtype(df[i]):
    print ("Média:", df[i].mean())
    print ("Mediana:", df[i].median())
    print ("Moda:", df[i].mode().tolist())
  else:
    print ("Moda:", df[i].mode().tolist())
```

```python
# Medidas de dispersão

# Interpretação:
# Variância/Desvio Padrão alto: Dados muito dispersos.
# Variância/Desvio Padrão baixo: Dados próximos da média.
# CV alto (> 30%) → Alta variabilidade relativa.
# CV baixo (< 10%) → Dados homogêneos.

for i in df.columns:
  if pd.api.types.is_numeric_dtype(df[i]):
    print(f"\n Variável: {i}")
    print ("Variância:", df[i].var())
    print ("Desvio Padrão:", df[i].std())
    if df[i].mean() != 0 and not np.isnan(df[i].mean()):
      print("Coeficiente de Variação (%):", (df[i].std() / df[i].mean()) * 100)
    else:
      print("Coeficiente de Variação (%): Não definido")
```

```python
for i in df.columns:
  if pd.api.types.is_numeric_dtype(df[i]):
    print (f"\n Variável: {i}")
    print ("Min:", df[i].min())
    print ("Máx:", df[i].max())
    print ("Quartis:\n", df[i].quantile([0.25, 0.5, 0.75]))
```

```python
print(df.isnull().sum())
```

```python
# IQR: Forma mais robusta de detectar um outlier
# IQR = Q3 - Q1

for i in df.columns:
  if pd.api.types.is_numeric_dtype(df[i]):
    q1 = df[i].quantile(0.25)
    q3 = df[i].quantile(0.75)
    iqr = q3 - q1
    outliers_iqr = df[(df[i] < (q1 - 1.5 * iqr)) | (df[i] > (q3 + 1.5 * iqr))]
    print (f"\n Variável: {i}")
    print ("Quantidade de Outliers (IQR):", outliers_iqr.shape[0])
```
