```python
import numpy as np
```

```python
data = np.loadtxt('./artificial2d.csv', delimiter=',')
X = data[:, :2]
y = data[:, -1]
print(X)
print(y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X[:,0], X[:,1], c=y)
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

# Sem regularização
# mdl = LogisticRegression(penalty=None)

# Com regularização
mdl = LogisticRegression(penalty='l2', C=1000)


mdl.fit(X, y)
y_pred = mdl.predict(X)
# Usando conjunto de treino só para exemplificar

print(y_pred)
print(y)
```

```python
from sklearn.inspection import DecisionBoundaryDisplay

DecisionBoundaryDisplay.from_estimator(mdl, X, response_method='predict')
plt.scatter(X[:,0], X[:,1], c=y)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y, y_pred)
print(cm)

vp = cm[0, 0]
print(vp)

fn = cm[0, 1]
print(fn)

fp = cm[1, 0]
print(fp)

vn = cm[1, 1]
print(vn)
```
