```python
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
df = np.loadtxt("artificial1d.csv", delimiter=',', skiprows=1)
```

```python
X = df[:, 0]
y = df[:, 1]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
car_2 = PolynomialFeatures(degree=2)
X = X.reshape(-1, 1)
X_pol_2 = car_2.fit_transform(X)
```

```python
modelo2 = LinearRegression()
modelo2.fit(X_pol_2, y)
y_pol_2 = modelo2.predict(X_pol_2)
```

```python
plt.scatter(X, y, label = 'Y(X)')
plt.plot(X, y_pol_2, color='red', label = 'ajuste')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
```
